Método que a partir de la palabra ambigua, o mejor dicho del nombre del archivo del corpus `word` para la palabra ambigua, obtiene cuáles son los posibles **sentidos de la palabra ambigua**:

```python
def senses(word):
    """
    This takes a target word from senseval-2 and it returns the list of possible senses for the word
    """
    return list(set(
        s
        for instance in senseval.instances(word)
        for s in instance.senses
    ))
```

```python
from nltk.corpus import senseval
from collections import defaultdict

def analizar_sentidos(palabra):
    instancias = senseval.instances(palabra)
    conteo = defaultdict(int)
    ejemplos = {}

    for instancia in instancias:
        for sentido in instancia.senses:
            conteo[sentido] += 1
            if sentido not in ejemplos:
                contexto = ' '.join([w for (w, _) in instancia.context])
                ejemplos[sentido] = contexto

    return conteo, ejemplos

def imprimir_tabla_analisis(palabra):
    conteo, ejemplos = analizar_sentidos(palabra)
    print(f"\n{'='*30} {palabra} {'='*30}")
    print(f"{'Sentido':<10} | {'# de instancias':<15} | {'Ejemplo de contexto':<50}")
    print("-"*100)
    for sentido in sorted(conteo):
        ejemplo = ejemplos[sentido][:50]
        print(f"{sentido:<10} | {conteo[sentido]:<15} | {ejemplo}")
    print("-"*100)

# Ejecutar para hard y serve
imprimir_tabla_analisis('hard.pos')
imprimir_tabla_analisis('serve.pos')
```

* ¿Cuántas instancias hay en el corpus para cada uno de los sentidos de las palabras ambiguas «hard» y «serve»? Es decir, cuantas oraciones hay en el corpus etiquetadas con cada uno de los sentidos.

Método que obtiene todas las **instancias de un posible sentido de la palabra ambigua** a partir de la lista de instancias de una palabra `instances` y el nombre del sentido palabra ambigua `sense`:

```python
def sense_instances(instances, sense):
    """
    This returns the list of instances in instances that have the sense `sense`
    """
    return [
        instancia
        for instancia in instances
        if sense in instancia.senses
    ]
```

```python
# Obtener las instancias del corpus
hard_instances = senseval.instances('hard.pos')
serve_instances = senseval.instances('serve.pos')

# Ejemplo: obtener todas las instancias con sentido HARD1
hard1 = sense_instances(hard_instances, 'HARD1')
serve6 = sense_instances(serve_instances, 'SERVE6')

# Mostrar cuántas instancias hay
print(f"Instancias con sentido HARD1: {len(hard1)}")
print(f"Instancias con sentido SERVE6: {len(serve6)}")
```

* En el contexto, las palabras ambiguas pueden aparecer en diferentes formas gramaticales. Por ejemplo, en el caso de la palabra ambigua *«hard»*, esta aparece tanto la forma base, el adjetivo *«hard»* como en comparativo *«harder»* y como en superlativo *«hardest»*. ¿Qué formas gramaticales aparecen en el contexto para cada una de las palabas ambiguas *«hard»* y *«serve»*?

```python
def formas_gramaticales(instances, base_word):
    formas = set()
    for instance in instances:
        for elemento in instance.context:
            word = elemento[0] if isinstance(elemento, tuple) else elemento
            if word.lower().startswith(base_word.lower()):
                formas.add(word.lower())
    return sorted(formas)

```

```python
formas_hard = formas_gramaticales(hard_instances, "hard")
formas_serve = formas_gramaticales(serve_instances, "serve")
```

Método que muestra las diferentes **versiones de la palabra ambigua** a partir del nombre del archivo del corpus `word` para la palabra ambiguala palabra ambigua:

```python
def tokens(word):
    """
    This takes a target word from senseval-2 and it returns the list of possible tokens for the word
    """
    token = set()
    for instance in senseval.instances(word):
        for elemento in instance.context:
            palabra = elemento[0] if isinstance(elemento, tuple) else elemento
            if palabra.lower().startswith(word.split('.')[0]):
                token.add(palabra.lower())
    return list(token)
```

- Para la eliminación del conjunto de palabras no útiles del vocabulario se puede usar un código parecido al que se indica a continuación. Debes tener en cuenta que **en el código faltaría añadir las palabras que has identificado en la Parte 1** de este laboratorio como las diferentes formas gramaticales de las palabras ambiguas.

```python
import nltk
nltk.download('stopwords')
```

```python
import string

OTHER_WORDS = ["''", '"', "'d", "'ll", "'m", "'re", "'s", "'t", "'ve", '--', '000',
               '1', '2', '3', '4', '5', '6', '8', '10', '15']

STOPWORDS_SET = set(stopwords.words('english')).union(set(string.punctuation), set(OTHER_WORDS))
```

- Ejemplo de vocabulario para un tamaño de 6. Por ejemplo, si se quiere entrenar un clasificador que permita identificar los diferentes sentidos de la palabra «hard» y se utilizan para entrenar y validar el modelo las instancias etiquetadas para esta palabra, la bolsa de palabras en el caso de considerar las seis palabras más frecuentes (m=6) sería la presentada anteriormente `['time', 'would', 'get', 'work', 'find', 'make']`.

Método que devuelve el **vocabulario** formado por las `m` palabras más frecuentes en el contexto de una palabra ambigua a partir de un conjunto de instancias `instances` y el conjunto de palabras no útiles `stopwords` (Nota: Recuerda que se deben eliminar las diferentes formas gramaticales de la palabra ambigua):

```python
def extract_vocab(instances, stopwords=STOPWORDS_SET, m=250):
    """
    Given a list of senseval instances, return a list of the m most frequent words that
    appears in its context (i.e., the sentence with the target word in), output is in order
    of frequency.
    """
    return [w for w,f in extract_vocab_frequency(instances,stopwords,m)]
```

```python
def extract_vocab_frequency(instances, stopwords=STOPWORDS_SET, m=250):
    """
    Dada una lista de instancias de senseval, devuelve una lista de las m palabras más frecuentes
    (excluyendo stopwords y formas gramaticales de la palabra ambigua).
    El resultado es una lista de tuplas (palabra, frecuencia), ordenada de mayor a menor frecuencia.
    """
    frecuencia = {}

    for instance in instances:
        palabras_en_contexto = set()  # usamos set para contar solo una vez por instancia
        for elemento in instance.context:
            palabra = elemento[0] if isinstance(elemento, tuple) else elemento
            palabra = palabra.lower()
            if palabra.isalpha() and palabra not in stopwords:
                palabras_en_contexto.add(palabra)

        for palabra in palabras_en_contexto:
            frecuencia[palabra] = frecuencia.get(palabra, 0) + 1

    # Ordenamos por frecuencia descendente
    vocab_ordenado = sorted(frecuencia.items(), key=lambda x: x[1], reverse=True)

    # Retornamos solo los m más frecuentes
    return vocab_ordenado[:m]
```

Utiliza un diccionario en Python para guardar el conjunto de características. La clave del diccionario serán las palabras del vocabulario y el valor debe ser un booleano para indicar la aparición o no de las palabras en el contexto. Por ejemplo, en el vector de características `{'contains(time)': True, 'contains(would)': False, 'contains(get)': False, 'contains(work)': False, 'contains(find)': False, 'contains(make)': False}` una de las claves del diccionario es `'contains(time)'` y su valor es `True`  lo que indica que en el contexto de la palabra ambigua aparece la palabra *«time»*.

- Debes mostrar el vector de características resultante para una de las instancias del corpus.

- **Importante:** En el cómputo del vector de características basado en las palabras vecinas debes utilizar como contexto la oración completa donde aparece la palabra ambigua. Es decir, todas las palabras que forman la oración guardada en el campo  `context` de la instancia.

Método que devuelve el conjunto de características basado en palabras vecinas para una instancia `instance` a partir de un vocabulario `vocab` (Nota: el parámetro `dist` no debes usarlo):

```python
def wsd_caracteristicas_palabras_vecinas(instance, stopwords_set=STOPWORDS_SET, number=250, distance=None):
    features = {}

    context = [w[0].lower() if isinstance(w, tuple) else w.lower() for w in instance.context]

    # Crear lista de palabras contextuales sin stopwords
    content_words = [w for w in context if w not in stopwords_set]

    # Seleccionar las n palabras más frecuentes
    for w in content_words[:number]:
        features[f'contains({w})'] = True

    return features
```

- **Importante:** Debes tener en cuenta los posibles casos en los que la palabra ambigua aparezca al principio o final de la frase, ya que en esas instancias no vas a poder obtener una secuencia de palabras de longitud n. Por ejemplo, para la instancia cuyo contexto es: `[('some', 'DT'), ('hard', 'JJ'), ('choices', 'NNS'), ('had', 'VBD'), ('to', 'TO'), ('be', 'VB'), ('made', 'VBN'), …]` si n=2 deberías obtener el siguiente vector de características: `{'previous(some)': True, 'next(choices had)': True}`.

- **Nota:** aunque no es imprescindible para realizar esta actividad de laboratorio, puedes utilizar las funcionalidades para trabajar con n-gramas que ofrece NLTK. Estas se pueden importar utilizando el siguiente comando:

> *from nltk import ngrams*

Método que devuelve el conjunto de características de colocación para una instancia `instance` usando los n-gramas anterior y posterior donde la longitud de secuencia es de `dist` palabras (Nota: el parámetro `vocab` no debes usarlo):

```python
def wsd_caracteristicas_colocacion(instance, vocab=None, dist=2):
    features = {}

    # Extraer contexto en minúsculas
    context = [w.lower() if isinstance(w, str) else w[0].lower() for w in instance.context]

    # Buscar manualmente la palabra ambigua que empiece con "hard"
    target_word = next((w for w in context if w.startswith("hard")), None)

    if target_word is None:
        print(f"No se encontró ninguna palabra que comience con 'hard' en el contexto: {context}")
        return features

    # Buscar índice
    target_index = None
    for i, word in enumerate(context):
        if word == target_word:
            target_index = i
            break

    if target_index is None:
        print(f"No se encontró la palabra ambigua '{target_word}' en el contexto: {context}")
        return features

    # Palabras anteriores y posteriores
    prev_words = context[max(0, target_index - dist):target_index]
    next_words = context[target_index + 1:target_index + 1 + dist]

    # Crear bigramas anteriores
    for i in range(len(prev_words)):
        bigram = '_'.join(prev_words[i:])
        features[f'previous({bigram})'] = True

    # Crear bigramas posteriores
    for i in range(len(next_words)):
        bigram = '_'.join(next_words[:i+1])
        features[f'next({bigram})'] = True

    return features
```

- **Clasificador basado en las palabras vecinas.** Con el conjunto de datos de entrenamiento, entrena un clasificador para «hard» que use como características el conjunto basado en las **palabras vecinas** cuyo código has implementado en la parte 2 de este laboratorio. - **Clasificador basado en características de colocación.** Con el conjunto de datos de entrenamiento, entrena un clasificador para «hard» que use como conjunto de características las de colocación cuyo código has implementado en la parte 2 de este laboratorio. Para definir la **ventana de contexto** utiliza la secuencia de dos palabras que ocurren antes de la palabra ambigua y la secuencia de dos palabras que ocurren después de esta **(n=2)**.

```python
from nltk.classify import accuracy, NaiveBayesClassifier
from nltk import ConfusionMatrix
import random

def wsd_clasificador(word, features_func, stopwords_list=STOPWORDS_SET, number=250, vocab=None, distance=2, errors=False, confusion_matrix=False):

    # Filtrar instancias para la palabra ambigua específica
    instances = [inst for inst in senseval.instances(word)]

    # Mezclar para garantizar aleatoriedad
    random.seed(42)
    random.shuffle(instances)

    # Dividir en 80% train y 20% test
    split_point = int(0.8 * len(instances))
    train_data = instances[:split_point]
    test_data = instances[split_point:]

    # Construir conjuntos de entrenamiento y prueba con features
    if features_func == wsd_caracteristicas_palabras_vecinas:
        train_set = [(features_func(inst, stopwords_list, number), inst.senses[0]) for inst in train_data]
        test_set  = [(features_func(inst, stopwords_list, number), inst.senses[0]) for inst in test_data]
    elif features_func == wsd_caracteristicas_colocacion:
        train_set = [(features_func(inst, vocab, distance), inst.senses[0]) for inst in train_data]
        test_set  = [(features_func(inst, vocab, distance), inst.senses[0]) for inst in test_data]
    else:
        train_set = [(features_func(inst), inst.senses[0]) for inst in train_data]
        test_set  = [(features_func(inst), inst.senses[0]) for inst in test_data]

    # Entrenar el clasificador
    classifier = NaiveBayesClassifier.train(train_set)

    # Evaluar accuracy
    acc = accuracy(classifier, test_set)
    print('Accuracy: %6.4f' % acc)

    # Mostrar errores si se solicita
    if errors:
        print('\nErrores:')
        errores = []
        for i, (featureset, gold) in enumerate(test_set):
            guess = classifier.classify(featureset)
            if guess != gold:
                instance = test_data[i]
                errores.append((instance, guess))

                context_words = [w[0] if isinstance(w, tuple) else w for w in instance.context]
                target_word = instance.word.lower()
                target_index = None
                for j, w in enumerate(context_words):
                    if isinstance(w, tuple):
                        w_clean = w[0].lower()
                    else:
                        w_clean = w.lower()
                    if w_clean == target_word:
                        target_index = j
                        break

                if target_index is not None:
                    start_index = max(0, target_index - 2)
                    end_index = min(len(context_words), target_index + 3)
                    context_snippet = ' '.join(context_words[start_index:end_index])
                    print(f"{i:>2}) {context_snippet}")
                    print(f"   Predicted: {guess}, Actual: {gold}")
                else:
                    print(f"{i:>2}) [Context not found] Predicted: {guess}, Actual: {gold}")

        return errores  # Retornar lista de errores

    # Mostrar matriz de confusión si se solicita
    if confusion_matrix:
        gold_labels = [label for (_, label) in test_set]
        predicted_labels = [classifier.classify(fs) for (fs, _) in test_set]
        cm = ConfusionMatrix(gold_labels, predicted_labels)
        print('\nMatriz de confusión:')
        print(cm)

    return classifier  # Retornar el clasificador si no se solicitaron errores
```
